**If Else Statements**


```python
age=18
if(age>=18):
  print("Minor")
else:
  print("Adult")
```

**If Else Elseif(Elif)**

```python
mark=90
if(mark>=90):
  print("O+")
elif(mark>=80):
  print("A+")
elif(mark>=70):
  print("B+")
elif(mark>=60):
  print("C+")
elif(mark>=50):
  print("D+")
else:
  print("Fail")
```
