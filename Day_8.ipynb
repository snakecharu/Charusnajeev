```python
# prompt: create a csv file with two cols,two rows

import csv

data = [
    ['col1', 'col2'],
    ['row1_val1', 'row1_val2'],
    ['row2_val1', 'row2_val2']
]

with open('my_data.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerows(data)

print("CSV file 'my_data.csv' created successfully.")
```
