**Part-2 in a Function**


**Default Parameters**

```python
def hello(name):
  print(f"Hello {name}!")
hello("Charu")
```

```python
#Default Parameters
def hello(name="Guest"):
  print(f"Hello {name}!")
hello()
```

**keyword Function**

```python
def add(x=10,y=10):
  print(x+y)
add(5,5)
add()
```

**Scope and Global Variables**

```python
x=10
def hell():
  print(x)
hell()
```

```python
x=10
def hello(y):
  global x
  print(x,y)
hello(20)
```

```python
x=10
def hello(x):
  x=20
  print(x)
hello(19)
```

**Nested Functions**

```python
def hello(name):
  print(f"hello {name}!")
def give(city,name):
  hello(name)
  print(f"Welcome to {city}")
give("Mumabi","Charu")
```
