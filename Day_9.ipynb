**Dictionary in Python**


**Creating dictionary**

```python
#dictionary-key_value pairs
student={"name":"Charu","age":19,"gender":"female"}
print(student)
```

**Accessing a dictionary**

```python

student["gender"]
```

```python
student["age"]
```

```python
student["name"]
```

**Looping through a dictionary**

```python
for i in student:
  print(i)
```

```python
for i in student.values():
  print(i)
```

```python
for i,j in student.items():
  print(i,j)
```

```python
for i in student:
  if i=="name":
    print(student[i])
```

```python
if "age" in student:
  print("true")
else:
  print("false")
```
