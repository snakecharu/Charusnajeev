**List in Comprehension**

```python
#without comprehension
marks=[90,80,70,60]
for i in marks:
  print(i/2)
```

```python
#with comprehension
marks=[90,80,70,60]
marks=[i/2 for i in marks]
print(marks)

#or
marks=[90,80,70,60]
print([i/2 for i in marks])
```
